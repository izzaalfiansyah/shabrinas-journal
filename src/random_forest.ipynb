```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv('../dataset/dataset_kematangan_tapai_v2.csv')
```

```python
input_columns = ['jam', 'kelembaban', 'kadar_gas']
output_columns = ['status_kematangan']
selection_columns = input_columns + output_columns

data = data.loc[:, selection_columns]

label_encoder = LabelEncoder()
scaler = StandardScaler()
```

```python
data['status_kematangan'] = label_encoder.fit_transform(data.loc[:, 'status_kematangan'])
data.head()
```

```python
x = scaler.fit_transform(data.loc[:, input_columns].values)
y = data.loc[:, output_columns].to_numpy().flatten()

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=0)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, accuracy_score, classification_report

model = RandomForestClassifier()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

acc_score = accuracy_score(y_test, y_pred)
f1_score = f1_score(y_test, y_pred, average='macro')
class_report = classification_report(y_test, y_pred)
print(f"Accuracy Score: {acc_score}")
print(f"F1 Score: {f1_score}")
print(class_report)
```
