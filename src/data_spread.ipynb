```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('../dataset/dataset_kematangan_tapai_v2.csv')
data.info()
```

```python
data.head()
```

```python
data = data.loc[:, ['jam', 'suhu', 'kelembaban', 'kadar_gas', 'status_kematangan']]
palette = {'belum matang': 'dodgerblue', 'matang': 'green', 'terlalu matang': 'red'}

ax = sns.scatterplot(data=data, x='jam', y='kadar_gas', hue='status_kematangan', palette=palette)
ax.set_title('Kadar gas berdasarkan jam')
```

```python
ax = sns.scatterplot(data=data, x='suhu', y='kadar_gas', hue='status_kematangan', palette=palette)
ax.set_title("Kadar gas berdasarkan suhu")
```

```python
ax = sns.scatterplot(data=data, x='kelembaban', y='kadar_gas', hue='status_kematangan', palette=palette)
ax.set_title("Kadar gas berdasarkan kelembaban")
```

```python
sns.boxplot(data=data, x='kadar_gas', y='status_kematangan', hue='status_kematangan', palette=palette)
```
