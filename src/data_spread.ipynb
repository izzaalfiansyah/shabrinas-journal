```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('../dataset/dataset_kematangan_tapai_v2.csv')
data.info()
```

```python
data.head()
```

```python
counts = data['status_kematangan'].value_counts()
palette = {'belum matang': '#3498db', 'matang': '#2ecc71', 'terlalu matang': '#e74c3c'}

sns.barplot(counts, palette=palette.values())
```

```python
data = data.loc[:, ['jam', 'suhu', 'kelembaban', 'kadar_gas', 'status_kematangan']]
correlations = data[['jam', 'suhu', 'kelembaban', 'kadar_gas']].corr()

sns.heatmap(correlations, annot=True, cmap='Blues')
```

```python
for feature in ['jam', 'suhu', 'kelembaban']:
    plt.figure()
    ax = sns.scatterplot(data=data, x=feature, y='kadar_gas', hue='status_kematangan', palette=palette)
    plt.show()
```

```python
for feature in ['kadar_gas', 'jam', 'suhu', 'kelembaban']:
    plt.figure()
    sns.boxplot(data=data, x=feature, y='status_kematangan', hue='status_kematangan', palette=palette)
    plt.show()
```
